```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('itachi.jpg', 0)
```

---

Histogram Equalization

```python
equ = cv2.equalizeHist(image)
```

```python
res = np.hstack((image, equ))
```

```python
plt.title("Histogram Equalization")
plt.imshow(cv2.cvtColor(res, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

---

Display Histogram

```python
hist = cv2.calcHist([image], [0], None, [256], [0, 256])
```

```python
plt.plot(hist)
plt.show()
```

```python
hist = cv2.calcHist([res], [0], None, [256], [0, 256])
```

```python
plt.plot(hist)
plt.show()
```

---

Histogram Matching

```python
from skimage import exposure
from skimage.exposure import match_histograms
import matplotlib.image as mpimg
```

```python
image1 = cv2.imread('image.jpg', 0)
image2 = cv2.imread('aot.png', 0)
```

```python
hist1 = cv2.calcHist([image1], [0], None, [256], [0, 256])
hist2 = cv2.calcHist([image2], [0], None, [256], [0, 256])
```

```python
matched = exposure.match_histograms(image1, image2, channel_axis=-1)
hist3 = cv2.calcHist([matched], [0], None, [256], [0, 256])
```

```python
plt.imshow(matched)
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title("Hist of image 1")
plt.plot(hist1)

plt.subplot(1, 3, 2)
plt.title("Hist of image 2")
plt.plot(hist2)

plt.subplot(1, 3, 3)
plt.title("Hist of Matched")
plt.plot(hist3)

plt.show()
```
