```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('image.jpg')
```

---

Negative Image

```python
negative = np.max(image) - image
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Negative")
plt.imshow(cv2.cvtColor(negative, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

---

Log Transformation

$$ s = c * log(r + 1) $$

```python
c = 255 / np.log(1.0 + np.max(image))
log_image = np.array(c * np.log(image + 1), dtype=np.uint8)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Log Image")
plt.imshow(cv2.cvtColor(log_image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

---

Power law Transformation

```python
gamma_val = 4.0
```

```python
result = np.array(c * image ** gamma_val, dtype=np.uint8)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Gamma Transformation")
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

---

Thresholding

- Binary Thresholding

```python
ret, thresh = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Threshold")
plt.imshow(cv2.cvtColor(thresh, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

- Binary Inverse Thresholding

```python
ret, thresh = cv2.threshold(image, 70, 255, cv2.THRESH_BINARY_INV)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Binary Inverse Threshold")
plt.imshow(cv2.cvtColor(thresh, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

- Truncate Thresholding

```python
ret, thresh = cv2.threshold(image, 70, 255, cv2.THRESH_TRUNC)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Truncate Threshold")
plt.imshow(cv2.cvtColor(thresh, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

- Threshold to Zero

```python
ret, thresh = cv2.threshold(image, 70, 255, cv2.THRESH_TOZERO)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title("image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Thresh to Zero")
plt.imshow(cv2.cvtColor(thresh, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```
