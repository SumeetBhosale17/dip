```python
import cv2
import matplotlib.pyplot as plt
```

```python
image1 = cv2.imread('image.jpg')

image2 = cv2.imread('aot.png')
```

---

- Addition of images

```python
addition = cv2.add(image1, image2)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title("image 1")
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("image 2")
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("addition")
plt.imshow(cv2.cvtColor(addition, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

---

- Subtraction of Images

```python
subtracted = cv2.subtract(image1, image2)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title("image 1")
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("image 2")
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Subtracted")
plt.imshow(cv2.cvtColor(subtracted, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

- Logical Operations

-   - Bitwise AND

```python
bitwise_and = cv2.bitwise_and(image2, image1)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title("image 1")
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("image 2")
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Bitwise And")
plt.imshow(cv2.cvtColor(bitwise_and, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

-   - OR

```python
bitwise_or = cv2.bitwise_or(image2, image1)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title("image 1")
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("image 2")
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Bitwise OR")
plt.imshow(cv2.cvtColor(bitwise_or, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

-   - XOR

```python
bitwise_xor = cv2.bitwise_xor(image2, image1)
```

```python
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title("image 1")
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("image 2")
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Bitwise XOR")
plt.imshow(cv2.cvtColor(bitwise_xor, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```
