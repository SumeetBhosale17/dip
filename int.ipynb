```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('image.jpg')
```

```python
img[100, 150]
```

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(R, cmap='Reds')

plt.subplot(1, 3, 2)
plt.imshow(G, cmap='Greens')

plt.subplot(1, 3, 3)
plt.imshow(B, cmap='Blues')

plt.show()
```

```python
h, w, _ = img.shape
```

```python
top_right = img[:h//2, w//2:]
bottom_right = img[h//2:, w//2:]
bottom_left = img[h//2:, :w//2]
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(top_right)
plt.title("Top Right")

plt.subplot(1, 3, 2)
plt.imshow(bottom_right)
plt.title("Bottom Right")

plt.subplot(1, 3, 3)
plt.imshow(bottom_left)
plt.title("Bottom Left")

plt.show()
```
