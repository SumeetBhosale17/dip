## Image Segmentation

```python
from skimage import data
from skimage import filters
from skimage.color import rgb2gray
import matplotlib.pyplot as plt
```

```python
coffee = data.coffee()
gray_coffee = rgb2gray(coffee)
```

```python
plt.figure(figsize=(15, 15))

for i in range(10):
    threshold = i * 0.1

    binary = (gray_coffee > threshold) * 1

    plt.subplot(5, 2, i+1)
    plt.imshow(binary, cmap='gray')
    plt.title(f"Threshold > {threshold:.1f}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
otsu_threshold = filters.threshold_otsu(gray_coffee)
otsu_binary = (gray_coffee > otsu_threshold) * 1
```

```python
niblack_threshold = filters.threshold_niblack(gray_coffee)
niblack_binary = (gray_coffee > niblack_threshold) * 1
```

```python
sauvola_threshold = filters.threshold_sauvola(gray_coffee)
sauvola_binary = (gray_coffee > sauvola_threshold) * 1
```

```python
plt.figure(figsize=(12, 10))

plt.subplot(2, 3, 1)
plt.imshow(gray_coffee, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(otsu_binary, cmap="gray")
plt.title(f"Otsu\nThreshold = {otsu_threshold:.4f}")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(niblack_binary, cmap="gray")
plt.title("Niblack Thresholding")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(sauvola_threshold, cmap="gray")
plt.title("Sauvola Threshold Values")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(sauvola_binary, cmap="gray")
plt.title("Sauvola Thresholding")
plt.axis("off")

plt.tight_layout()
plt.show()
```
