- **Resize using Percentage**

```python
height, width, _ = src.shape
```

```python
percent = 50

w, h = int(width * percent / 100), int(height * percent / 100)
```

```python
resize = cv2.resize(src, (w, h))
resize.shape
```

---

- **Image Pyramid using OpenCV**

```python
import matplotlib.pyplot as plt
```

```python
layer = src.copy()

for i in range(4):
    plt.subplot(2, 2, i+1)
    layer = cv2.pyrDown(layer)
    plt.imshow(layer)
    cv2.imshow("str(i)", layer)
    cv2.waitKey(0)
cv2.destroyAllWindows()
```

- **Image quantization**

```python
from PIL import Image
import matplotlib.pyplot as plt
```

```python
img = Image.open(r"image.jpg")
img1 = img.quantize(10)
plt.imshow(img1)
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img)

plt.subplot(1, 2, 2)
plt.title("Quantized")
plt.imshow(img1)

plt.show()
```
