```python
import cv2
import matplotlib.pyplot as plt
```

```python
image1 = cv2.imread('image.jpg')
image2 = cv2.imread('Itachi.jpg')
```

```python
bitwise_xor = cv2.bitwise_xor(image1, image2)
```

```python
plt.figure(figsize=(20, 5))

plt.subplot(1, 3, 1)
plt.title("Image 1")
plt.imshow(cv2.cvtColor(image1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 2")
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Bitwise XOR")
plt.imshow(cv2.cvtColor(bitwise_xor, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

```python
image = cv2.imread("image.jpg")
```

```python
import numpy as np
```

```python
c = 255.0 / (np.log(1.0 + np.max(image)))
```

```python
log_image = np.array(c * np.log(image + 1), dtype=np.uint8)
```

```python
plt.figure(figsize=(20, 5))

plt.subplot(1, 2, 1)
plt.title("Image")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Log Image")
plt.imshow(cv2.cvtColor(log_image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```
