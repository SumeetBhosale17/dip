```python
import cv2
```

```python
from IPython.display import Image
Image('image.jpg')
```

1(a) Aim:  To create a program to display the image using read and show operation.

```python
image = cv2.imread('image.jpg')
if image is None:
    raise FileNotFoundError("Image not found")
cv2.imshow('Image', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

1(b) Aim: To create a program to rotate an image.

```python
(h, w) = image.shape[:2]
center = (w/2, h/2)
angle = 45
scale = 1.0

M = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(image, M, (w, h))

cv2.imshow("Rotated Image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rotate_90 = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
cv2.imshow('Rotated 90', rotate_90)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

1(c) Aim : To create a program to flip an image.

```python
flipped_vertically = cv2.flip(image, 0)
flipped_horizontally = cv2.flip(image, 1)
flipped_both = cv2.flip(image, -1)

cv2.imshow('Original', image)
cv2.imshow('Flipped Vertically', flipped_vertically)
cv2.imshow('Flipped Horizontally', flipped_horizontally)
cv2.imshow('Flipped Both', flipped_both)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

1(d) Aim : To create a program to transpose an image.

```python
transposed = cv2.transpose(image)

cv2.imshow('Original', image)
cv2.imshow('Transposed', transposed)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

1(e) Aim : To create a program to convert original image to grayscale.

```python
gray_scale = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Original', image)
cv2.imshow('Gray Scale', gray_scale)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
