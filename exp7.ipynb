```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('image.jpg', 0)
```

---

```python
m, n = image.shape
```

```python
mask = np.ones([3, 3], dtype=int)
mask = mask / 9
```

```python
image_new = np.zeros([m, n])
```

```python
for i in range(1, m-1):
    for j in range(1, n-1):
        temp = (
        image[i-1, j-1] * mask[0, 0] + image[i-1, j] * mask[0, 1] + image[i-1, j+1] * mask[0, 2] + 
        image[i, j-1] * mask[1, 0] + image[i, j] * mask[1, 1] + image[i, j+1] * mask[1, 2] + 
        image[i+1, j-1] * mask[2, 0] + image[i + 1, j] * mask[2, 1] + image[i+1, j+1] * mask[2, 2]
        )

        image_new[i, j] = temp
```

```python
image_new = image_new.astype(np.uint8)
cv2.imwrite('blurred.tif', image_new)
```
